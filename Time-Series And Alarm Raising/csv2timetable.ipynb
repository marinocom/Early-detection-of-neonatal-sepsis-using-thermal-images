```python
import pandas as pd
import matplotlib.pyplot as plt
import os

def path2graphic(path):
    df=pd.read_csv(path, parse_dates=["timestamp"])
    # df_filtered=pd.read_csv(path_filtered, parse_dates=["timestamp"])

    df["date"]=df["timestamp"].dt.date
    # df_filtered["date"]=df_filtered["timestamp"].dt.date

    points=["Zona1", "codo_izq", "codo_drc", "rodilla_izq", "rodilla_drc", "mano_izq", "mano_drc", "pie_izq", "pie_drc"]

    plt.figure(figsize=(14, 6))

    for point in points:
        if point in df.columns:
            plt.plot(df["timestamp"], df[point], label=point, marker='o')

    plt.title("Evolución de Temperatura todos los puntos")
    plt.xlabel("Tiempo")
    plt.ylabel("Temperatura (°C)")
    plt.xticks(rotation=45)
    plt.grid(True)
    plt.legend()

    plt.show()
    plt.close()

    print("\nGraficas por zona \n")

    for point in points:
        if point in df.columns:
            plt.figure(figsize=(10, 4))
            plt.plot(df["timestamp"], df[point], marker='o', color='tab:blue')
            plt.title(f"Evolución de temperatura: {point}")
            plt.xlabel("Tiempo")
            plt.ylabel("Temperatura (°C)")
            plt.grid(True)
            plt.xticks(rotation=45)
            
            plt.show()
            plt.close()

    print("\nGraficas por zona y dia \n")

    for point in points:
        if point in df.columns:
            for (date, group) in (df.groupby("date")):
                plt.figure(figsize=(10, 4))
                plt.plot(group["timestamp"], group[point], marker='o', color='tab:orange', alpha=1)
                # plt.plot(group2["timestamp"], group2[point], marker='o', color='tab:blue', alpha=0.5)
                plt.title(f"{point} - {date}")
                plt.xlabel("Hora")
                plt.ylabel("Temperatura (°C)")
                plt.grid(True)
                plt.xticks(rotation=45)

                plt.show()
                plt.close()
    
    for point in points:
        if point == "Zona1":
            plt.figure(figsize=(10, 4))
            plt.plot(df["timestamp"], df[point], marker='o', color='tab:orange')
            
        elif point == "mano_izq":
            plt.plot(df["timestamp"], df[point], marker='o', color='tab:blue', alpha=0.1)

    plt.title(f"Evolución de temperatura: {point}")
    plt.xlabel("Tiempo")
    plt.ylabel("Temperatura (°C)")
    plt.grid(True)
    plt.xticks(rotation=45)
    plt.show()
    plt.close()
```
